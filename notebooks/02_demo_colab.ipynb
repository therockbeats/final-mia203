## Checkpoint del grupo
Descargue best.pt desde el enlace de weights/README.md.

```python
from google.colab import files
uploaded_weights = files.upload()
WEIGHTS = next((Path(n).resolve() for n in uploaded_weights if n.endswith(".pt")), None)
if WEIGHTS is None:
    raise ValueError("Seleccione el checkpoint best.pt")
```

```python
uploaded_image = files.upload()
IMAGE = next((Path(n).resolve() for n in uploaded_image if Path(n).suffix.lower() in {".jpg", ".jpeg", ".png"}), None)
if IMAGE is None:
    raise ValueError("Seleccione una imagen")
```

```python
import torch
from IPython.display import display, Image
EXPERIMENT = "M2" # utilizar best.pt del experimento seleccionado
DEVICE = "0" if torch.cuda.is_available() else "cpu"
subprocess.run([sys.executable, "scripts/demo.py", "--image", str(IMAGE),
    "--weights", str(WEIGHTS), "--experiment", EXPERIMENT, "--output", "/content/deteccion.jpg", "--device", DEVICE], check=True)
display(Image(filename="/content/deteccion.jpg"))
```

La imagen de salida muestra la clase, la confianza y la ubicación de cada detección.
